# Laboratory 1- Introduction

Creating the Dataset

```python
data = {
    "User": [
        "Alice", "Alice", "Alice",
        "Bob", "Bob", "Bob",
        "Carol", "Carol", "Carol",
        "David", "David",
        "Emma", "Emma"
    ],
    "Movie": [
        "Matrix", "Titanic", "Avatar",
        "Matrix", "Avatar", "Inception",
        "Titanic", "Inception", "Avatar",
        "Matrix", "Inception",
        "Titanic", "Avatar"
    ],
    "Rating": [
        5, 3, 4,
        4, 5, 5,
        4, 5, 3,
        5, 4,
        5, 4
    ]
}
df = pd.DataFrame(data)
df
```

## Task 1.1
Find:

• Number of users \
• Number of movies \
• Total number of ratings

```python
print("Number of users:", df['User'].nunique())
```

```python
print("Number of movies:", df['Movie'].nunique())
```

```python
print("Total number of ratings:", len(df))
```

## Task 1.2
Find the average rating:

```python
print("Average rating:", df['Rating'].mean().round(2))
```

## Task 2.1
Compute average rating per movie:

```python
print("Average rating per movie:")
avg_rating_per_movie = df.groupby('Movie')['Rating'].mean().round(2)
avg_rating_per_movie
```

## Task 2.2
Sort movies by rating:

```python
avg_rating_per_movie.to_frame().sort_values(by="Rating", ascending=False)
```

## Task 2.3
Find the best movie:

```python
max_rating = avg_rating_per_movie.max()
best_movies = avg_rating_per_movie[avg_rating_per_movie == max_rating].index.tolist()
best_rating = avg_rating_per_movie[avg_rating_per_movie == max_rating].values.tolist()
```

```python
print("Best movie(s):", best_movies, "with rating:", best_rating)
```

## Task 3.1
Find users who rated Matrix ≥ 4:

```python
df[(df["Movie"] == "Matrix") & (df["Rating"] >= 4)]
```

## Task 3.2
Extract these users:

```python
users = df[(df["Movie"] == "Matrix") & (df["Rating"] >= 4)].drop_duplicates(subset=["User"])["User"].to_list()
```

```python
users
```

## Task 3.3
Find movies rated by these users:

```python
films_rated_by_users = df[df["User"].isin(users)]
films_rated_by_users
```

## Task 3.4
Remove Matrix:

```python
films_rated_by_users_without_matrix = films_rated_by_users[films_rated_by_users["Movie"] != "Matrix"]
films_rated_by_users_without_matrix
```
